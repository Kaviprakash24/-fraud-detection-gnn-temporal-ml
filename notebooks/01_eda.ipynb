# Exploratory Data Analysis
This notebook investigates the PaySim dataset.
It explores transaction distributions, fraud percentages, transaction amounts, and temporal activity.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Create figures directory
os.makedirs('../results/figures', exist_ok=True)

# Load data
df = pd.read_csv('../data/raw/paysim dataset.csv')
print(f"Dataset shape: {df.shape}")
```

```
# Fraud vs Non-Fraud Distribution
fraud_counts = df['isFraud'].value_counts()
fraud_pct = df['isFraud'].mean() * 100

print(f"Fraud count: {fraud_counts[1]}")
print(f"Normal count: {fraud_counts[0]}")
print(f"Fraud Percentage: {fraud_pct:.3f}%")

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='isFraud')
plt.title(f'Fraud vs Normal Transactions (Fraud: {fraud_pct:.3f}%)')
plt.savefig('../results/figures/fraud_distribution.png')
plt.show()
```

```
# Transaction Type Distribution & Fraud Rate by Type
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='type', hue='isFraud')
plt.title('Transaction Type Distribution by Fraud Status')
plt.yscale('log')
plt.ylabel('Count (Log Scale)')
plt.savefig('../results/figures/type_distribution_log.png')
plt.show()

# Fraud rate by type
fraud_by_type = df.groupby('type')['isFraud'].mean() * 100
print("Fraud percentage by transaction type:")
print(fraud_by_type)
```

```
# Transaction Amount Distribution
plt.figure(figsize=(12, 6))
sns.histplot(df[df['amount'] < 500000]['amount'], bins=50, kde=False)
plt.title('Transaction Amount Distribution (Amounts < 500k)')
plt.savefig('../results/figures/amount_distribution.png')
plt.show()

# Fraud Amount Distribution
plt.figure(figsize=(12, 6))
sns.histplot(df[df['isFraud']==1]['amount'], bins=50, color='red')
plt.title('Fraudulent Transaction Amount Distribution')
plt.savefig('../results/figures/fraud_amount_distribution.png')
plt.show()
```

```
# Transactions over time (Step = 1 hour)
# Group by step
step_counts = df.groupby('step').size()
fraud_step_counts = df[df['isFraud'] == 1].groupby('step').size()

plt.figure(figsize=(15, 6))
step_counts.plot(label='All Transactions')
fraud_step_counts.plot(label='Fraud Transactions', color='red')
plt.title('Transactions over Time (Step)')
plt.xlabel('Step (Hour)')
plt.ylabel('Count')
plt.legend()
plt.savefig('../results/figures/transactions_over_time.png')
plt.show()
```

```
# Sender and Receiver Activity
top_senders = df['nameOrig'].value_counts().head(10)
top_receivers = df['nameDest'].value_counts().head(10)

print("Top 10 Senders (by count):")
print(top_senders)

print("\nTop 10 Receivers (by count):")
print(top_receivers)
```
